- `print()`：画面に表示する

### 実行する前に予想しよう

最初の呼び出しでは30が表示されます。その後、`result = add(10, 20)` の `result` には何が入るでしょうか。

````python
def add(a, b):
    answer = a + b
    print(answer)

add(10, 20)
````

````python
result = add(10, 20)

print(result)
````

### 結果を確認しよう

## 11. APIから天気を取得する

APIは、別のサービスが持つデータや機能を、プログラムから利用するための窓口です。ここではOpen-MeteoへHTTPのGETリクエストを送り、東京付近の現在の気温と天気コードを取得します。

処理の流れは次のとおりです。

1. URLでAPIの場所を指定する
2. `params` 辞書で緯度・経度・取得項目などを指定する
3. `requests.get()` で要求を送る
4. `response.json()` で応答をPythonの辞書・リストとして扱える形にする

### 実行する前に予想しよう

- `params["current"]` では何を要求していますか。

> **実行環境**　このセルにはインターネット接続と `requests` ライブラリが必要です。実行時刻やAPIの状態によって結果は変わります。

````python
import requests

url = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude": 35.68,
    "longitude": 139.76,
    "current": "temperature_2m,weather_code",
    "timezone": "Asia/Tokyo",
}

response = requests.get(url, params=params)

data = response.json()

print(data)
````

## 12. 天気コードをアイコンへ変換する

Open-Meteoの `weather_code` はWMOの天気解釈コードに基づく数値です。条件分岐を使い、サイネージで分かりやすいアイコンへ変換します。

次の対応表を満たす関数を作ります。

| weather_code | 天気 | 表示するアイコン |
|---:|---|---|
| 0 | 晴れ | ☀️ |
| 1, 2 | 晴れ時々曇り | 🌤️ |
| 3 | 曇り | ☁️ |
| 45, 48 | 霧 | 🌫️ |
| 51, 53, 55 | 霧雨 | 🌧️ |
| 61, 63, 65 | 雨 | 🌧️ |
| 71, 73, 75 | 雪 | ❄️ |
| 80, 81, 82 | にわか雨 | 🌦️ |
| 95, 96, 99 | 雷雨 | ⛈️ |
| 上記以外 | 不明 | ❓ |

### 関数を完成させよう

関数名は `weather_code_to_icon` とします。最初の2つの条件を例として用意してあります。対応表を見ながら、この続きに `elif` と `else` を追加してください。

- 1つのコードとの比較：`weather_code == 0`
- 複数候補との比較：`weather_code in [1, 2]`
- 表にないコード：最後の `else` で処理
- アイコンを関数の外へ返す：`return`

> **注意**　関数名と引数名は変更せず、関数の中を完成させます。

````python
def weather_code_to_icon(weather_code):
    if weather_code == 0:
        return "☀️"
    elif weather_code in [1, 2]:
        return "🌤️"

    # 対応表を見ながら、この下に elif と else を追加する
````

### 実行結果を予想しよう

関数を完成させたら、次の確認用コードがどの順番で何を表示するか、対応表を見ながら予想してください。

確認用コードは変更せず、そのまま使います。予想を書き終えたら実行して確かめましょう。

````python
test_codes = [0, 2, 3, 45, 53, 61, 71, 80, 95, 100]

for code in test_codes:
    print(code, weather_code_to_icon(code))
````

### 結果を確認しよう

- すべての行が対応表どおりのアイコンになりましたか。
- `None` が表示されたコードは、対応する `elif` または `else` が不足しています。
- 異なるアイコンが表示された場合は、条件の数値または `return` の内容を確認します。
- インデントがそろっているかも確認しましょう。
