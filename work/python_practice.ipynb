wether_codeから取得したコードをもとに天気アイコンを表示する関数を作成しなさい。
WMO（世界気象機関）のWeather interpretation codesを基にしたコード

| weather_code | 天気 | アイコン |
|---:|---|---|
| 0 | 晴れ | ☀️ |
| 1, 2 | 晴れ時々曇り | 🌤️ |
| 3 | 曇り | ☁️ |
| 45, 48 | 霧 | 🌫️ |
| 51, 53, 55 | 霧雨 | 🌧️ |
| 61, 63, 65 | 雨 | 🌧️ |
| 71, 73, 75 | 雪 | ❄️ |
| 80, 81, 82 | にわか雨 | 🌦️ |
| 95, 96, 99 | 雷雨 | ⛈️ |

````python
def weather_code_to_icon(weather_code):
    if weather_code == 0:
        return "☀️"
    elif weather_code in [1, 2]:
        return "🌤️"
    elif weather_code == 3:
        return "☁️"
    elif weather_code in [45, 48]:
        return "🌫️"
    elif weather_code in [51, 53, 55]:
        return "🌧️"
    elif weather_code in [61, 63, 65]:
        return "🌧️"
    elif weather_code in [71, 73, 75]:
        return "❄️"
    elif weather_code in [80, 81, 82]:
        return "🌦️"
    elif weather_code in [95, 96, 99]:
        return "⛈️"
    else:
        return "❓"
        
````

````python
weather_icone = weather_code_to_icon(weather_code)
````

````python
print(weather_icone)
````
